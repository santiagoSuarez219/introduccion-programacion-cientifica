```python
import datetime

UNIDADES_VALIDAS = ("ug/m3","mg/m3","kg/m3")

# Molde
class Medicion:
  # Atributos o caracteristicas
  # Metodo Constructor
  def __init__(self, valor, unidad, estacion):
    if valor < 0:
      raise ValueError("el valor de la medicion no debe ser negativo")
    if unidad not in UNIDADES_VALIDAS:
      raise ValueError("unidad invalida")
    self.valor = valor
    self.unidad = unidad
    self.estacion = estacion
    self.fecha = datetime.datetime.now()                   # En java es el this

  # Metodos o comportamientos de la clase: que acciones puede hacer?
  def convertir_miligramos_m3(self):
    if self.unidad == "mg/m3":
      return self.valor
    return self.valor * 1000

  def __str__(self):
    return f"MEDIDA: {self.valor} {self.unidad} - ESTACION: {self.estacion}"
```

```python
# TODO: Definir una clase estacion
class Estacion:
  # TODO: atributos: codigo, municipio, barrio, esta_disponible, mediciones
  def __init__(self, codigo, municipio, barrio, esta_disponible):
    self.codigo = codigo
    self.municipio = municipio
    self.barrio = barrio
    self.esta_disponible = esta_disponible
    self.mediciones = [] # Lista vacia

  # TODO: metodo: agregar_medicion: agrega una medicion a las mediciones de la estacion
  def agregar_medicion(self, medicion):
    self.mediciones.append(medicion)

  # TODO: metodo: promedio de esas mediciones
  def promedio(self):
    total = 0
    for medicion in self.mediciones:
      total = total + medicion.valor
    promedio = total / len(self.mediciones)
    return promedio

  # TODO: modificar el metodo __str___ para que muestre el codigo de la estacion, municipio, esta_disponible
  def __str__(self):
    return f"CODIGO: {self.codigo} MUNICIPIO: {self.municipio} DISPONIBLE: {self.esta_disponible}"
```
